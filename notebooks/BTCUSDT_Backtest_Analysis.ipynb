# BTCUSDT London Dynamic Breakout
A portfolio notebook reviewing a five-year Python backtest with 0.05% current-equity risk per entry.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
trades = pd.read_csv(ROOT / 'results' / 'trades.csv', parse_dates=['entry_time_utc', 'exit_time_utc'])
equity = pd.read_csv(ROOT / 'results' / 'daily_equity.csv', index_col=0, parse_dates=True)['equity_usd']
summary = json.loads((ROOT / 'results' / 'summary.json').read_text())
summary['metrics']
```

## Equity and drawdown

```python
drawdown = equity / equity.cummax() - 1
fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
equity.plot(ax=axes[0], title='BTCUSDT Equity Curve', color='#0b7285')
axes[0].set_ylabel('Equity (USD)')
axes[1].fill_between(drawdown.index, drawdown.values * 100, 0, color='#d1495b')
axes[1].set_ylabel('Drawdown %')
plt.tight_layout();
```
